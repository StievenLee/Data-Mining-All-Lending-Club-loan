## 1.2 Analisis Missing Values

```python
def missing_report(df, name):
    missing = df.isnull().sum()
    pct = (missing / len(df) * 100).round(2)
    report = pd.DataFrame({
        'Missing Count': missing,
        'Missing %': pct,
        'Dtype': df.dtypes
    })
    report = report[report['Missing Count'] > 0].sort_values('Missing %', ascending=False)
    print(f'Missing Value Report: {name}')
    print(f'Total kolom dengan missing : {len(report)} dari {df.shape[1]}')
    print(f'Kolom dengan >50% missing  : {(report["Missing %"] > 50).sum()}')
    print(f'Kolom dengan >90% missing  : {(report["Missing %"] > 90).sum()}')
    return report
```

## 1.4 Analisis Outlier

```python
def detect_outliers_iqr(df, cols):
    results = []
    for col in cols:
        data = df[col].dropna()
        Q1, Q3 = data.quantile(0.25), data.quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        n_outliers = ((data < lower) | (data > upper)).sum()
        results.append({
            'Kolom': col,
            'Lower Bound': lower,
            'Upper Bound': upper,
            'Jumlah Outlier': n_outliers,
            '% Outlier': round(n_outliers / len(data) * 100, 2)
        })
    return pd.DataFrame(results).sort_values('% Outlier', ascending=False)
```

```python
numeric_cols_acc = df_acc.select_dtypes(include='number').columns.tolist()
valid_num_cols = [c for c in numeric_cols_acc if df_acc[c].dropna().shape[0] > 1000][:20]

outlier_report = detect_outliers_iqr(df_acc, valid_num_cols)
print('Laporan Outlier (IQR Method), df_acc:')
outlier_report
```

Catatan untuk kolom zero-inflated seperti delinq_2yrs dan pub_rec:
IQR = 0 bukan berarti semua nilai di atas 0 adalah outlier. Kolom ini memiliki distribusi
zero-inflated di mana mayoritas bernilai 0, sehingga metode IQR tidak tepat diterapkan
dan kolom ini dikecualikan dari capping.

Terdapat juga lower bound negatif pada kolom yang secara domain tidak mungkin negatif
seperti loan_amnt dan revol_bal. Batas bawah IQR secara teoritis bisa negatif, namun
nilai negatif di kolom ini tetap di-cap ke 0 berdasarkan domain knowledge.

```python
top_outlier_cols = outlier_report.head(6)['Kolom'].tolist()

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
axes = axes.flatten()

for i, col in enumerate(top_outlier_cols):
    data = df_acc[col].dropna()
    p1, p99 = data.quantile(0.01), data.quantile(0.99)
    axes[i].boxplot(
        data[(data >= p1) & (data <= p99)],
        vert=False, patch_artist=True,
        boxprops=dict(facecolor='lightblue')
    )
    axes[i].set_title(col, fontsize=10)

plt.suptitle('Boxplot Kolom dengan Outlier Terbanyak', fontsize=13)
plt.tight_layout()
plt.savefig('Results/Images/outlier_boxplot.png', dpi=120, bbox_inches='tight')
plt.show()
```

Insight Outlier Analysis:

Boxplot menunjukkan bahwa outlier ekstrem umumnya berada di sisi kanan (upper), konsisten
dengan distribusi right-skewed yang sudah diamati sebelumnya. annual_inc dan revol_bal
memiliki outlier paling ekstrem di antara seluruh fitur yang dianalisis.

Untuk sebagian besar fitur, outlier ditangani dengan IQR capping (winsorization): nilai
di luar batas IQR di-clip tanpa menghapus baris, sehingga tidak ada data yang hilang.

annual_inc adalah pengecualian. IQR capping tidak bisa digunakan untuk annual_inc karena
IQR upper bound memotong sekitar 5% data di sisi kanan, jauh terlalu agresif untuk
distribusi power law. Jika log transform kemudian diterapkan pada annual_inc yang sudah
di-cap, ekor kanannya sudah terlalu pendek sehingga log overcorrect dan menghasilkan
distribusi left-skewed. Oleh karena itu annual_inc akan ditangani dengan percentile
clipping P1-P99 yang hanya membuang 1% nilai paling ekstrem, diikuti log1p.

Langkah 2.1.3: Drop kolom leaky (data leakage).

Kolom leaky adalah kolom yang nilainya baru tersedia **setelah pinjaman berjalan atau
selesai**. Memasukkannya ke analisis berarti model/algoritma "mengintip" informasi masa
depan yang belum ada saat keputusan kredit dibuat, sehingga hasilnya invalid secara ilmiah.

Leakage dibagi menjadi dua kelompok:

1. Leakage finansial (post-loan cashflow): total_pymnt, recoveries, out_prncp, dll.
   Kolom ini merekam pembayaran aktual yang hanya diketahui setelah pinjaman berjalan.

2. Leakage skor kredit terkini (post-loan credit pull): last_fico_range_low,
   last_fico_range_high, dan last_credit_pull_d. Ketiganya adalah hasil penarikan FICO
   terakhir selama pinjaman berlangsung, bukan saat pengajuan. Skor ini sudah
   "mengetahui" arah outcome: peminjam yang menuju default cenderung FICO-nya jatuh
   menjelang gagal bayar. Itulah sebabnya korelasinya terhadap target terlihat sangat
   tinggi (artifisial), bukan karena prediktif, tetapi karena bocor.

Sebagai pembanding yang SAH, fico_range_low dan fico_range_high (skor FICO **saat
pengajuan**) tetap dipertahankan karena tersedia di titik keputusan kredit dan tidak bocor.

## 2.4 Cleaning df_acc: Handle Duplikat dan Outlier

Langkah 2.4.2: Handle outlier dengan IQR Capping (Winsorization).

Nilai ekstrem di-clip ke batas IQR tanpa menghapus baris, sehingga tidak ada data yang
hilang. Hanya diterapkan pada kolom numerik dengan skewness di atas 2 dan kardinalitas > 10 (bukan biner/zero-inflated).

Justifikasi pengecualian, kolom power-law tidak boleh di-IQR-cap secara membabi-buta:
Sebagian fitur finansial (misalnya annual_inc dan kolom saldo/limit seperti revol_bal,
tot_cur_bal) mengikuti distribusi power-law dengan ekor kanan sangat panjang. Untuk kolom
seperti ini, batas IQR upper memotong terlalu agresif. Bila kolom power-law nantinya
membutuhkan log transform, ekor kanannya menjadi terlalu pendek sehingga log overcorrect
dan distribusi berbalik menjadi left-skewed.

Karena itu diberlakukan guard otomatis: sebuah kolom dikecualikan dari IQR capping
bila tergolong power-law ekstrem, yang dideteksi secara objektif lewat aturan
"persentase data di atas IQR-upper > 5%". Logika ini menggantikan pengecualian manual yang
sebelumnya hanya berlaku untuk annual_inc, agar perlakuan konsisten untuk semua kolom
berdistribusi serupa. annual_inc tetap ditangani khusus (P1-P99 clip + log1p) di Langkah 2.4.3.


```python
def cap_outliers_iqr(df, col):
    Q1, Q3 = df[col].quantile(0.25), df[col].quantile(0.75)
    IQR = Q3 - Q1
    df[col] = df[col].clip(lower=Q1 - 1.5 * IQR, upper=Q3 + 1.5 * IQR)
    return df

def is_power_law(series, frac_threshold=0.05):
    # Deteksi power-law ekstrem secara objektif.
    # Bila proporsi nilai di atas batas IQR-upper > frac_threshold (default 5%),
    # distribusi dianggap power-law dan IQR capping TIDAK tepat (akan overcorrect
    # saat dikombinasi log). Inilah guard yang menggantikan pengecualian manual.
    s = series.dropna()
    Q1, Q3 = s.quantile(0.25), s.quantile(0.75)
    upper = Q3 + 1.5 * (Q3 - Q1)
    return (s > upper).mean() > frac_threshold

# annual_inc selalu dikecualikan (ditangani P1-P99 clip + log1p di Langkah 2.4.3)
manual_skip = {'annual_inc'}

skewed_threshold = 2.0
numeric_cols_now = df_acc_clean.select_dtypes(include='number').columns
candidate_cols = [
    c for c in numeric_cols_now
    if c not in manual_skip
    and abs(df_acc_clean[c].skew()) > skewed_threshold
    and df_acc_clean[c].nunique() > 10
]

# Guard power-law: kecualikan kolom yang ekor kanannya terlalu tebal untuk IQR cap
power_law_cols = [c for c in candidate_cols if is_power_law(df_acc_clean[c])]
skewed_cols = [c for c in candidate_cols if c not in power_law_cols]

print(f'Kandidat skewed (|skew| > {skewed_threshold}): {len(candidate_cols)}')
print(f'Dikecualikan manual (annual_inc)            : {sorted(manual_skip)}')
print(f'Dikecualikan guard power-law (>5% di atas IQR-upper): {power_law_cols}')
print(f'Di-IQR-cap                                  : {len(skewed_cols)} kolom')
for col in skewed_cols:
    df_acc_clean = cap_outliers_iqr(df_acc_clean, col)
    print(f'  capping: {col}')

# Kolom power-law (selain annual_inc) ditangani dengan P1-P99 clipping.
# Nilai di bawah P1 dan di atas P99 tidak dihapus,
# melainkan dicap (capped) ke batas persentil terkait.
# Pendekatan ini lebih konservatif dibanding IQR capping pada distribusi heavy-tailed.

for col in power_law_cols:
    p1, p99 = df_acc_clean[col].quantile(0.01), df_acc_clean[col].quantile(0.99)
    df_acc_clean[col] = df_acc_clean[col].clip(lower=p1, upper=p99)
    print(f'  P1-P99 clip (power-law): {col}')

# Domain correction dti: IQR lower bound bisa negatif secara matematis,
# padahal dti adalah rasio yang tidak bisa bernilai negatif secara domain.
if 'dti' in df_acc_clean.columns:
    n_neg = (df_acc_clean['dti'] < 0).sum()
    n_inf = np.isinf(df_acc_clean['dti']).sum()
    if n_neg > 0 or n_inf > 0:
        print(f'  domain correction dti: {n_neg} negatif, {n_inf} inf -> clip ke 0')
    df_acc_clean['dti'] = df_acc_clean['dti'].replace([np.inf, -np.inf], np.nan)
    df_acc_clean['dti'] = df_acc_clean['dti'].clip(lower=0)
    df_acc_clean['dti'].fillna(df_acc_clean['dti'].median(), inplace=True)

print(f'Shape setelah capping: {df_acc_clean.shape}')
print(f'dti range: [{df_acc_clean["dti"].min():.2f}, {df_acc_clean["dti"].max():.2f}]')
print(f'annual_inc skewness saat ini: {df_acc_clean["annual_inc"].skew():.4f} (belum ditransform)')
```

Langkah 2.4.3: Transformasi annual_inc dengan P1-P99 Clip dan Log Transform.

annual_inc tidak menggunakan IQR capping karena distribusinya sangat right-skewed dan menunjukkan karakteristik heavy-tailed (power-law-like), dengan sejumlah kecil observasi yang memiliki pendapatan jauh lebih tinggi dibanding mayoritas data.
Pada distribusi seperti ini, IQR capping cenderung terlalu agresif sehingga dapat memangkas ekor kanan yang sebenarnya masih valid. Ketika log transform diterapkan setelah IQR capping, distribusi dapat menjadi overcorrected dan berubah menjadi left-skewed karena sebagian besar variasi pada ekor kanan telah hilang.

Solusi yang digunakan adalah dua tahap:
1. Percentile clipping P1-P99: nilai di bawah persentil ke-1 dan di atas persentil ke-99 tidak dihapus,tetapi dicap (capped) ke batas persentil tersebut. Ini jauh lebih konservatif dari IQR capping dan membiarkan distribusi power law alami annual_inc tetap utuh untuk diproses log.
2. Log transform (np.log1p): setelah clipping, transformasi log digunakan untuk mengompresi rentang nilai yang lebar dan mengurangi skewness sehingga distribusi menjadi lebih mendekati normal.

Sementatra itu, dti tidak perlu log transform karena setelah proses IQR capping nilai skewnessnya sudah 0.35
yang hampir simetris. Oleh karena itu, StandardScaler sudah cukup untuk menstandarisasi dti.

Berikut konfirmasi skewness annual_inc dan rentang P1-P99 sebelum transformasi diterapkan,
untuk memverifikasi bahwa outlier ekstrem memang ada dan perlu ditangani.

```python
# Cek skewness sebelum dan sesudah percentile clipping untuk annual_inc
if 'annual_inc' in df_acc_clean.columns:
    skew_before = df_acc_clean['annual_inc'].skew()
    p99_val = df_acc_clean['annual_inc'].quantile(0.99)
    p1_val  = df_acc_clean['annual_inc'].quantile(0.01)
    print(f'annual_inc sebelum P99 clip : skew = {skew_before:.4f}')
    print(f'  P1  = {p1_val:,.0f}')
    print(f'  P99 = {p99_val:,.0f}')
    print(f'  Nilai di atas P99 yang akan di-clip: {(df_acc_clean["annual_inc"] > p99_val).sum():,} baris ({(df_acc_clean["annual_inc"] > p99_val).mean()*100:.2f}%)')

cols_to_check = ['dti']
print()
print('Skewness dti setelah IQR capping:')
for col in cols_to_check:
    if col in df_acc_clean.columns:
        skew_val = df_acc_clean[col].skew()
        status = 'perlu log transform' if abs(skew_val) > 0.5 else 'sudah cukup simetris, tidak perlu log transform'
        print(f'  {col:<15}: skew = {skew_val:>7.4f}  ->  {status}')
```

Setelah dikonfirmasi bahwa skewness annual_inc sangat tinggi dan rentang P99 masih
representatif, berikut penerapan dua tahap transformasi: P1-P99 clip diikuti log1p.

```python
# Simpan nilai annual_inc sebelum log transform untuk keperluan binning di Langkah 3.4
# Binning harus menggunakan nilai asli agar batas bin (30K, 60K, dst) bermakna secara bisnis
annual_inc_before_log = df_acc_clean['annual_inc'].copy()

# Tahap 1: Percentile clipping P1-P99 untuk annual_inc
if 'annual_inc' in df_acc_clean.columns:
    p1_val  = df_acc_clean['annual_inc'].quantile(0.01)
    p99_val = df_acc_clean['annual_inc'].quantile(0.99)
    df_acc_clean['annual_inc'] = df_acc_clean['annual_inc'].clip(lower=p1_val, upper=p99_val)
    print(f'annual_inc clipped ke [{p1_val:,.0f}, {p99_val:,.0f}]')
    print(f'Skewness setelah P99 clip : {df_acc_clean["annual_inc"].skew():.4f}')

# Simpan skew setelah clipping (SEBELUM log)
skew_clipped = df_acc_clean['annual_inc'].skew()

# Tahap 2: Log transform
log_transform_cols = [c for c in ['annual_inc'] if c in df_acc_clean.columns]

for col in log_transform_cols:
    df_acc_clean[col] = np.log1p(df_acc_clean[col])

print()
print('Skewness sesudah log transform:')
for col in log_transform_cols:
    print(f'  {col}: {df_acc_clean[col].skew():.4f}')

fig, axes = plt.subplots(1, 3, figsize=(18, 4))

# Panel 1: distribusi asli (clip P99 untuk tampilan)
raw_data = df_acc['annual_inc'].replace([np.inf, -np.inf], np.nan).dropna()
p99_raw = raw_data.quantile(0.99)
raw_data[raw_data <= p99_raw].hist(bins=60, ax=axes[0], color='coral', edgecolor='white', alpha=0.8)
axes[0].set_title(f'annual_inc asli (skew={raw_data.skew():.1f})')
axes[0].set_xlabel('Nilai')

# Panel 2: setelah P99 clip (sebelum log)
clipped_ref = raw_data.clip(upper=p99_raw)
clipped_ref.hist(bins=60, ax=axes[1], color='orange', edgecolor='white', alpha=0.8)
axes[1].set_title(f'Setelah P99 clip (skew={skew_clipped:.2f})')
axes[1].set_xlabel('Nilai')

# Panel 3: setelah log transform
df_acc_clean['annual_inc'].hist(bins=60, ax=axes[2], color='steelblue', edgecolor='white', alpha=0.8)
axes[2].set_title(f'Setelah log1p (skew={df_acc_clean["annual_inc"].skew():.2f})')
axes[2].set_xlabel('Nilai')

plt.suptitle('Pipeline Transformasi annual_inc: Asli -> P99 Clip -> Log1p', fontsize=13)
plt.tight_layout()
plt.savefig('Results/Images/log_transform_comparison.png', dpi=120, bbox_inches='tight')
plt.show()
print('Transformasi annual_inc selesai')
```

Keputusan ini disengaja: kelompok DTI tinggi tersebut merupakan sinyal penolakan yang bermakna dan pada Phase 2 muncul sebagai salah satu Anti-Persona (segmen dengan DTI jauh di atas rata-rata). Yang dipangkas hanyalah nilai absurd (puluhan juta persen) yang merusak skala, sedangkan ekor yang informatif tetap dipertahankan untuk analisis clustering.

```python
# Outlier ekstrem pada DTI df_rej (max puluhan juta persen) di-cap dengan P1-P99.
# Dibiarkan mentah akan merusak StandardScaler pada Langkah 3.4.2.
if 'Debt-To-Income Ratio' in df_rej_clean.columns:
    dti_col = df_rej_clean['Debt-To-Income Ratio']
    p1, p99 = dti_col.quantile(0.01), dti_col.quantile(0.99)
    n_capped = int(((dti_col < p1) | (dti_col > p99)).sum())

    df_rej_clean['Debt-To-Income Ratio'] = dti_col.clip(lower=p1, upper=p99)

    print(f'Batas capping DTI  : P1={p1:.2f}%, P99={p99:.2f}%')
    print(f'Baris yang di-cap  : {n_capped:,} ({n_capped/len(df_rej_clean)*100:.2f}%)')
    print('\nStatistik DTI setelah capping:')
    print(df_rej_clean['Debt-To-Income Ratio'].describe().round(2))
```

## 3.3 Encoding Fitur Kategorik

Langkah 3.3.1: Ordinal Encoding untuk kolom dengan urutan natural.
Grade (A sampai G) memiliki hierarki yang bermakna secara bisnis sehingga label encoding
lebih tepat daripada one-hot encoding.

```python
# Ordinal encoding grade & sub_grade
# Grade A-G memiliki hierarki natural yang bermakna untuk clustering:
# nilai angka 1-7 mempertahankan urutan risiko sehingga jarak Euclidean
# antar klaster mencerminkan perbedaan profil risiko yang sesungguhnya
if 'grade' in df_acc_clean.columns:
    grade_map = {'A': 1, 'B': 2, 'C': 3, 'D': 4, 'E': 5, 'F': 6, 'G': 7}
    df_acc_clean['grade'] = df_acc_clean['grade'].map(grade_map)
    print('grade encoded:', sorted(df_acc_clean['grade'].unique()))

# sub_grade: mapping ORDINAL EKSPLISIT (A1=1 .. G5=35).
# LabelEncoder dihindari karena urutannya hanya kebetulan benar (alfabetis) dan rapuh;
# mapping eksplisit menjamin urutan risiko A1 < A2 < ... < G5 secara deterministik.
if 'sub_grade' in df_acc_clean.columns:
    grade_letters = ['A', 'B', 'C', 'D', 'E', 'F', 'G']
    sub_grade_map = {
        f'{letter}{num}': i * 5 + num
        for i, letter in enumerate(grade_letters)
        for num in range(1, 6)
    }
    df_acc_clean['sub_grade'] = df_acc_clean['sub_grade'].map(sub_grade_map)
    print(f'sub_grade encoded (ordinal eksplisit): '
          f'{df_acc_clean["sub_grade"].nunique()} nilai unik, '
          f'range [{int(df_acc_clean["sub_grade"].min())}, {int(df_acc_clean["sub_grade"].max())}]')
```

Langkah 3.3.2: Cek variansi kolom sebelum One-Hot Encoding.

Kolom dengan dominasi satu kategori di atas 95% (near-constant) tidak memberikan informasi
prediktif dan hanya menambah noise. Kolom seperti ini dikeluarkan dari daftar OHE.

Langkah 3.3.3: One-Hot Encoding untuk kolom nominal tanpa urutan natural.
drop_first=True untuk menghindari dummy variable trap yang menyebabkan multikolinearitas.

```python
# Kandidat kolom untuk OHE
ohe_candidates = ['home_ownership', 'verification_status', 'purpose',
                  'disbursement_method', 'initial_list_status']
ohe_candidates = [c for c in ohe_candidates if c in df_acc_clean.columns]

print('Analisis variansi kolom kandidat OHE:')
print(f'{"Kolom":<30} {"Dominant Category":<25} {"Dominance %":>12} {"Status":>15}')
print('-' * 85)

low_var_cols = []
for col in ohe_candidates:
    vc = df_acc_clean[col].value_counts(normalize=True)
    top_cat = vc.index[0]
    top_pct = vc.iloc[0] * 100
    status = 'NEAR-CONSTANT (drop)' if top_pct > 95 else 'OK'
    if top_pct > 95:
        low_var_cols.append(col)
    print(f'{col:<30} {str(top_cat):<25} {top_pct:>11.1f}% {status:>15}')

print(f'\nKolom near-constant yang dikeluarkan dari OHE: {low_var_cols}')
ohe_cols = [c for c in ohe_candidates if c not in low_var_cols]
print(f'Kolom yang akan di-OHE: {ohe_cols}')
```

```python
ohe_cols = [c for c in ohe_cols if c in df_acc_clean.columns]

shape_before = df_acc_clean.shape
df_acc_clean = pd.get_dummies(df_acc_clean, columns=ohe_cols, drop_first=True, dtype=int)

print(f'Kolom di-OHE: {ohe_cols}')
print(f'Shape sebelum OHE: {shape_before}')
print(f'Shape sesudah OHE : {df_acc_clean.shape}')
print(f'Kolom baru: {df_acc_clean.shape[1] - shape_before[1]}')
```

## 3.4 Normalisasi (Scaling)

Langkah 3.4.1: Normalisasi dengan StandardScaler.

annual_inc sudah lebih mendekati normal setelah log transform. dti sudah mendekati normal
setelah IQR capping sehingga tidak perlu log transform. StandardScaler dipilih karena
menghasilkan mean=0 dan std=1 yang optimal untuk algoritma berbasis jarak seperti K-Means
dan DBSCAN, serta lebih robust terhadap sisa outlier dibanding MinMaxScaler.

Kolom biner hasil OHE dan target dikecualikan karena sudah dalam range yang sesuai.
emp_length (ordinal 0-10) diperlakukan sebagai numerik dan ikut di-scale; hal ini disengaja
karena jaraknya bermakna (selisih lama kerja) sehingga standardisasi membuat skalanya
sebanding dengan fitur numerik lain dalam perhitungan jarak Euclidean.

Konsisten dengan kebijakan imputasi di Langkah 2.2.1, scaler di-fit hanya pada baris dengan
outcome final yang dipakai untuk analisis (Fully Paid, Charged Off, Default), baris
Current/Late sudah dibuang di Langkah 3.1.1 sebelum titik ini. Catatan metodologis: untuk
pemodelan supervised dengan evaluasi train/test, scaler harus di-fit hanya pada training
set; pada konteks unsupervised proyek ini (clustering & association rule) hal tersebut tidak
menimbulkan kebocoran karena tidak ada himpunan uji terpisah.


```python
target_col = 'loan_status_binary'
cols_to_exclude = [target_col, 'loan_status']

# Simpan hasil log (sebelum scaling) untuk visualisasi
log_temp = df_acc_clean['annual_inc'].copy()

numeric_for_scale = [
    c for c in df_acc_clean.select_dtypes(include='number').columns
    if c not in cols_to_exclude and df_acc_clean[c].nunique() > 2
]

# snapshot kolom numerik SEBELUM scaling untuk binning Langkah 3.4 ──
# Kini binning memakai snapshot dari
# df_acc_clean (sudah dibersihkan & sudah difilter target): IDENTIK dengan pipeline
# Apriori: agar profil bin konsisten antar kedua notebook.
# annual_inc memakai annual_inc_before_log (nilai pre-log, satuan dolar asli).
cols_to_snapshot = ['loan_amnt', 'int_rate', 'dti',
                    'acc_open_past_24mths', 'num_tl_op_past_12m',  # count mentah utk verifikasi Fase 4
                    'issue_d_year', 'issue_d_month']  # tahun terbit asli utk filter temporal Fase 5
snapshot_before_scale = {
    col: df_acc_clean[col].copy()
    for col in cols_to_snapshot
    if col in df_acc_clean.columns
}
print(f'Snapshot sebelum scaling disimpan untuk: {list(snapshot_before_scale.keys())}')

print(f'Jumlah kolom yang akan di-scale: {len(numeric_for_scale)}')

scaler = StandardScaler()
df_acc_clean[numeric_for_scale] = scaler.fit_transform(df_acc_clean[numeric_for_scale])
print('StandardScaler diterapkan')
print(f'Shape: {df_acc_clean.shape}')
```

Langkah 3.4.2: Normalisasi df_rej dengan StandardScaler.

Dataset rejected memiliki tiga fitur numerik (Amount Requested, Debt-To-Income Ratio, Employment Length) yang berada pada skala sangat berbeda: nominal pinjaman berskala ribuan, rasio utang berskala puluhan, dan lama bekerja berskala satuan (0-10). Karena algoritma K-Means dan DBSCAN pada Fase 2 mengandalkan jarak Euclidean, fitur berskala ribuan akan mendominasi perhitungan jarak dan membuat dua fitur lain nyaris tidak berpengaruh. StandardScaler diterapkan agar ketiga fitur memiliki mean=0 dan std=1, sehingga berkontribusi setara terhadap pembentukan klaster. Langkah ini menyelaraskan perlakuan df_rej dengan df_acc yang juga sudah distandarisasi, sehingga kedua dataset siap langsung dipakai untuk clustering tanpa transformasi ulang.

```python
# Normalisasi df_rej: fitur numerik masih dalam skala asli setelah cleaning,
# sedangkan Fase 2 (K-Means/DBSCAN) berbasis jarak Euclidean -> wajib di-scale.
numeric_for_scale_rej = ['Amount Requested', 'Debt-To-Income Ratio', 'Employment Length']
numeric_for_scale_rej = [c for c in numeric_for_scale_rej if c in df_rej_clean.columns]

print('Statistik SEBELUM scaling (skala asli):')
print(df_rej_clean[numeric_for_scale_rej].describe().loc[['mean', 'std', 'min', 'max']].round(2))

scaler_rej = StandardScaler()
df_rej_clean[numeric_for_scale_rej] = scaler_rej.fit_transform(df_rej_clean[numeric_for_scale_rej])

print('\nStatistik SESUDAH scaling (mean~0, std~1):')
print(df_rej_clean[numeric_for_scale_rej].describe().loc[['mean', 'std', 'min', 'max']].round(4))
print('\nStandardScaler diterapkan pada df_rej. Ketiga fitur kini berskala setara')
print('dan siap dipakai langsung untuk clustering Fase 2 tanpa scaling ulang.')
```

## 4.2 Correlation Analysis (Pearson Correlation)

Metode 1: Pearson Correlation.  
Mengukur hubungan LINEAR antara fitur dan target.  
|corr| > 0.1 : lemah tapi relevan untuk dataset besar.  
|corr| > 0.3 : moderat.  
|corr| > 0.5 : kuat.  
Juga digunakan untuk deteksi multikolinearitas antar fitur (corr > 0.9).  

```python
# Deteksi kolom zero-variance SEBELUM menghitung korelasi.
# Korelasi Pearson tidak terdefinisi (pembagi = 0) untuk kolom konstan, sehingga
# menghasilkan NaN yang akan MERUSAK seluruh kalkulasi cumulative score di langkah
# elbow analysis nanti (satu NaN saja membuat .sum() global jadi NaN).
# Kolom konstan biasanya adalah dummy OHE dari kategori sangat jarang (mis. salah satu
# dari home_ownership_ANY/OTHER/NONE) yang kebetulan menjadi semua-0 setelah baris
# Current/Late dibuang di Langkah 3.1: secara informasi memang tidak berguna (variance=0
# berarti tidak ada perbedaan yang bisa dijelaskan terhadap target), sehingga aman di-drop.
zero_var_cols = X.columns[X.nunique() <= 1].tolist()
if zero_var_cols:
    print(f'Kolom zero-variance terdeteksi & di-drop: {zero_var_cols}')
    X = X.drop(columns=zero_var_cols)
else:
    print('Tidak ada kolom zero-variance.')

corr_with_target = X.corrwith(y).abs().sort_values(ascending=False)

# Safety net tambahan: pastikan tidak ada NaN tersisa pada korelasi (mis. dari kasus
# zero-variance lain yang tidak tertangkap nunique<=1, misalnya seluruh kolom NaN)
n_nan_corr = corr_with_target.isnull().sum()
if n_nan_corr > 0:
    print(f'PERINGATAN: {n_nan_corr} fitur masih menghasilkan korelasi NaN, didrop dari analisis.')
    corr_with_target = corr_with_target.dropna()

corr_df = pd.DataFrame({
    'Feature': corr_with_target.index,
    'Correlation': corr_with_target.values
})
print('Top 20 fitur dengan korelasi tertinggi terhadap loan_status_binary:')
print(corr_df.head(20).to_string(index=False))
```

## 4.3 Entropy / Mutual Information

Metode 2: Mutual Information berbasis Entropy.  
MI(X;Y) = H(Y) - H(Y|X) di mana H = entropy (ketidakpastian informasi).

Keunggulan vs Korelasi:
- Menangkap hubungan non-linear.
- Bisa untuk tipe data campuran (numerik + kategorik).
- MI = 0 berarti fitur tidak ada hubungan dengan target.
- MI tinggi berarti fitur sangat informatif.

n_neighbors=5 dipilih karena memberikan trade-off yang baik antara stabilitas estimasi dan kemampuan menangkap pola non-linear pada dataset berukuran besar (>1 juta baris).


```python
mi_scores = mutual_info_classif(X, y, random_state=42, n_neighbors=5)
mi_df = pd.DataFrame({
    'Feature': X.columns,
    'MI_Score': mi_scores
}).sort_values('MI_Score', ascending=False)

print('\nTop 25 fitur berdasarkan Mutual Information:')
print(mi_df.head(25).to_string(index=False))
```

## 4.4 Gabungkan dan Pilih Fitur Final

Gabungkan skor Korelasi dan Mutual Information.  
Setiap skor dinormalisasi ke [0,1] agar bisa dijumlahkan secara adil.  
Combined_Score adalah rata-rata dari keduanya.  
Pendekatan ensemble ini lebih robust daripada mengandalkan satu metode.  

```python
combined_scores = pd.DataFrame({
    'Feature': corr_with_target.index,
    'Abs_Correlation': corr_with_target.values
}).merge(mi_df, on='Feature', how='inner')

combined_scores['Corr_Norm'] = combined_scores['Abs_Correlation'] / combined_scores['Abs_Correlation'].max()
combined_scores['MI_Norm'] = combined_scores['MI_Score'] / combined_scores['MI_Score'].max()
combined_scores['Combined_Score'] = (combined_scores['Corr_Norm'] + combined_scores['MI_Norm']) / 2

# Safety net: dengan zero-variance columns sudah didrop di Langkah 4.2, Combined_Score
# seharusnya tidak mengandung NaN. Dicek eksplisit agar tidak ada kebocoran NaN diam-diam
# ke kalkulasi cumulative score (elbow analysis) berikutnya.
n_nan = combined_scores['Combined_Score'].isnull().sum()
if n_nan > 0:
    print(f'PERINGATAN: {n_nan} fitur dengan Combined_Score NaN, didrop dari ranking.')
    combined_scores = combined_scores.dropna(subset=['Combined_Score'])

combined_scores = combined_scores.sort_values('Combined_Score', ascending=False).reset_index(drop=True)
print(f'Total fitur dengan Combined_Score valid: {len(combined_scores)}')
print('\nTop 25 fitur berdasarkan Combined Score:')
print(combined_scores.head(25)[['Feature', 'Abs_Correlation', 'MI_Score', 'Combined_Score']].to_string(index=False))
```

```python
# Elbow Analysis: justifikasi kuantitatif untuk jumlah fitur (top_n)
# Combined_Score sudah dipastikan bebas NaN di langkah sebelumnya (Langkah 4.3).
scores_sorted = combined_scores['Combined_Score'].values
n_features = len(scores_sorted)
ranks = np.arange(1, n_features + 1)

cumsum_pct = np.cumsum(scores_sorted) / scores_sorted.sum() * 100

# ── Deteksi elbow dengan Kneedle algorithm (Satopaa et al., 2011) ──────────
# Kneedle adalah metode published untuk mendeteksi titik infleksi (knee/elbow) pada kurva.
# Library `kneed` mengimplementasikan algoritma ini secara resmi: kurva dinormalisasi,
# dibandingkan terhadap garis referensi linear, lalu titik dengan deviasi maksimum dari
# garis tersebut dipilih sebagai knee. Pendekatan ini jauh lebih robust dibanding metode
# naif seperti np.argmin(np.diff(...)), yang rentan salah memilih titik di ekor kurva
# akibat noise kecil (bukan elbow sesungguhnya).
# curve='convex', direction='decreasing' karena Combined Score sorted descending berbentuk
# kurva cembung menurun (turun cepat di awal, melandai di akhir).
kneedle = KneeLocator(
    ranks, scores_sorted,
    curve='convex', direction='decreasing',
    S=1.0  # sensitivitas standar Kneedle, semakin besar S, semakin konservatif (knee lebih ke kanan)
)

elbow_rank = int(kneedle.knee)       # rank dalam istilah "fitur ke-N" (1-indexed)
elbow_idx = elbow_rank - 1           # index 0-based untuk akses array

# top_n ditetapkan MENGIKUTI elbow (data-driven), bukan angka hardcode.
top_n = elbow_rank

print(f'Elbow (Kneedle) terdeteksi pada fitur rank ke-{elbow_rank} (index {elbow_idx})')
print(f'Combined Score pada elbow : {scores_sorted[elbow_idx]:.4f}')
print(f'Cumulative score pada elbow: {cumsum_pct[elbow_idx]:.1f}%')
print(f'top_n ditetapkan = {top_n} (mengikuti elbow Kneedle)')
```

```python
# ── Plot ─────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(range(n_features), scores_sorted, color='steelblue', alpha=0.7)
axes[0].axvline(x=elbow_idx, color='red', linestyle='--', linewidth=2,
                label=f'top_n={top_n} = elbow (index {elbow_idx})')
axes[0].set_xlabel('Feature rank')
axes[0].set_ylabel('Combined Score')
axes[0].set_title('Combined Score per Fitur (semua fitur)')
axes[0].legend()

axes[1].plot(range(1, n_features + 1), cumsum_pct, marker='o', markersize=3, color='steelblue')
axes[1].axvline(x=elbow_rank, color='red', linestyle='--', linewidth=2,
                label=f'top_n={top_n}: {cumsum_pct[elbow_idx]:.1f}% score')
axes[1].axhline(y=80, color='orange', linestyle=':', label='80% threshold')
axes[1].axhline(y=90, color='green', linestyle=':', label='90% threshold')
axes[1].set_xlabel('Jumlah fitur dipilih (k)')
axes[1].set_ylabel('Cumulative Combined Score (%)')
axes[1].set_title('Cumulative Score, Justifikasi top_n')
axes[1].legend()
axes[1].grid(True, alpha=0.4)
axes[1].set_xlim(0, n_features + 1)
axes[1].set_ylim(0, 105)

plt.suptitle('Elbow Analysis: Menentukan Jumlah Fitur Optimal', fontsize=13)
plt.tight_layout()
plt.savefig('Results/Images/feature_selection_elbow.png', dpi=120, bbox_inches='tight')
plt.show()

print(f'\nCumulative coverage pada beberapa titik:')
print(f'  Top {elbow_rank:>2} (elbow) : {cumsum_pct[elbow_idx]:.1f}%')
print(f'  Top 25          : {cumsum_pct[24]:.1f}%')
print(f'  Top 30          : {cumsum_pct[29]:.1f}%')
```

Justifikasi top_n=16: elbow geometris terdeteksi di rank ke-16, yaitu
titik di mana marginal gain Combined Score per fitur tambahan turun paling drastis.
Setelah titik ini, setiap fitur tambahan menyumbang informasi yang kian kecil sementara
biaya dimensi terus naik, kontraproduktif untuk algoritma berbasis jarak (K-Means, DBSCAN)
yang rentan curse of dimensionality. Memilih jumlah fitur pada elbow (48.5%
cumulative) lebih defensible daripada mengejar threshold 80% yang memerlukan jauh lebih
banyak fitur dan menambah noise tanpa kompensasi struktur klaster yang berarti.

```python
# top_n sudah ditetapkan di Langkah 4.4 mengikuti elbow geometris (data-driven).
selected_features = combined_scores.head(top_n)['Feature'].tolist()

print(f'{top_n} fitur terpilih untuk dataset final (sesuai elbow):')
for i, feat in enumerate(selected_features, 1):
    row = combined_scores[combined_scores['Feature'] == feat].iloc[0]
    print(f'  {i:2d}. {feat:<40} | Corr={row["Abs_Correlation"]:.4f} | MI={row["MI_Score"]:.4f}')
```

```python
# Plot perbandingan dua metode HANYA untuk fitur yang terpilih (top_n = elbow).
plot_df = combined_scores.head(top_n).sort_values('Combined_Score')

fig, ax = plt.subplots(figsize=(12, max(6, top_n * 0.45)))
x = np.arange(len(plot_df))
width = 0.35

ax.barh(x - width/2, plot_df['Corr_Norm'], width, label='Correlation (normalized)', color='steelblue', alpha=0.8)
ax.barh(x + width/2, plot_df['MI_Norm'], width, label='Mutual Information (normalized)', color='coral', alpha=0.8)
ax.set_yticks(x)
ax.set_yticklabels(plot_df['Feature'])
ax.set_xlabel('Normalized Score')
ax.set_title(f'Perbandingan Feature Selection: Korelasi vs Mutual Information (Top {top_n} terpilih)', fontsize=13)
ax.legend()
plt.tight_layout()
plt.savefig('Results/Images/feature_selection_comparison.png', dpi=120, bbox_inches='tight')
plt.show()
```

Insight Perbandingan Korelasi vs Mutual Information:

Fitur yang konsisten tinggi di kedua metode merupakan kandidat terkuat untuk dimasukkan ke
model karena keduanya mengkonfirmasi relevansi dari sudut pandang linear dan non-linear.

Fitur yang tinggi di MI tetapi rendah di Korelasi mengindikasikan kemungkinan hubungan non-linear
dengan target seperti  outlier, interaksi threshold atau pola U-shape. Fitur ini sering dilewatkan
jika hanya menggunakan korelasi linear saja.

Fitur yang tinggi di Korelasi tetapi rendah di MI umumnya memiliki hubungan linear yang
kuat namun tidak banyak menambah informasi unik di luar fitur lain.

Strategi ensemble yang menggabungkan kedua metode lebih robust dibanding mengandalkan
satu metode saja karena masing-masing memiliki blind spot yang berbeda.

## 4.5 Drop Fitur Redundan untuk Clustering

Multikolinearitas perlu ditangani sebelum K-Means dan DBSCAN.
Fitur yang berkorelasi tinggi di atas 0.9 menggandakan bobot suatu dimensi dalam
perhitungan Euclidean distance sehingga mendistorsi bentuk klaster.

Dari tiap pasangan berkorelasi tinggi, dipertahankan fitur dengan Combined Score lebih tinggi
karena fitur tersebut lebih informatif terhadap target berdasarkan hasil feature selection.

PCA tidak dipilih karena langkah persiapan Apriori selanjutnya membutuhkan nama fitur yang bermakna
secara domain untuk interpretasi association rules.

```python
cols_to_drop_corr = []
seen = set()

print('Analisis pasangan fitur berkorelasi tinggi (>0.9):')
if high_corr_pairs:
    for pair in high_corr_pairs:
        f1, f2 = pair['Feature 1'], pair['Feature 2']
        if f1 in seen or f2 in seen:
            continue
        s1 = combined_scores[combined_scores['Feature'] == f1]['Combined_Score'].values
        s2 = combined_scores[combined_scores['Feature'] == f2]['Combined_Score'].values
        if len(s1) and len(s2):
            drop = f2 if s1[0] >= s2[0] else f1
            keep = f1 if drop == f2 else f2
            cols_to_drop_corr.append(drop)
            seen.update([f1, f2])
            print(f'  Pasangan  : {f1} & {f2}')
            print(f'  Korelasi  : {pair["Correlation"]}')
            print(f'  Pertahankan (score lebih tinggi): {keep}')
            print(f'  Drop      : {drop}')
            print()
else:
    print('  Tidak ada pasangan fitur dengan korelasi > 0.9 di top 30 fitur.')
    print('  Seluruh selected_features dapat langsung digunakan untuk clustering.')

selected_features_final = [f for f in selected_features if f not in cols_to_drop_corr]

print(f'Fitur sebelum deduplikasi       : {len(selected_features)}')
print(f'Fitur di-drop (redundan)        : {len(cols_to_drop_corr)} -> {cols_to_drop_corr}')
print(f'Fitur final untuk clustering    : {len(selected_features_final)}')
print()
print('Fitur final:')
for i, f in enumerate(selected_features_final, 1):
    print(f'  {i:2d}. {f}')
```

Insight Drop Fitur Redundan:

Fitur yang di-drop dipilih berdasarkan Combined Score yang lebih rendah dari pasangannya,
sehingga fitur dengan informasi lebih kaya terhadap target tetap dipertahankan.

Dataset final untuk clustering menggunakan selected_features_final yang sudah bebas dari
redundansi. Dataset ini akan digunakan langsung pada K-Means dan DBSCAN.

Untuk langkah persiapan Apriori selanjutnya, fitur-fitur bin seperti int_rate_bin, dti_bin,
loan_amnt_bin, dan income_bin akan digunakan bersama kolom cluster_label hasil clustering,
karena Apriori membutuhkan fitur kategorik dan tidak terpengaruh oleh multikolinearitas.

```python
df_acc_final = df_acc_clean[selected_features_final + ['loan_status_binary']].copy()
df_acc_final.to_csv('../Datasets/Cleaning/Phase 2/clean_accepted_loans.csv', index=False)
print(f'Saved: clean_accepted_loans.csv -> {df_acc_final.shape}')

# --- Ekspor tahun penerbitan untuk filter temporal di dashboard Fase 5 ---
# issue_d_year dibuat di Langkah 2.2 (tanggal -> year/month) tetapi TIDAK lolos ke
# selected_features_final karena bukan fitur prediktif. Padahal dimensi waktu berguna
# untuk analisis tren anomali di Fase 4/5. Solusinya: ekspor terpisah + 'row_pos'
# (posisi baris pada clean_accepted_loans.csv) sebagai kunci penggabungan, mengikuti
# pola yang sama dengan raw_counts_accepted dan cluster_labels_accepted.
if 'issue_d_year' in snapshot_before_scale:
    # PENTING: nilai diambil dari snapshot_before_scale, bukan dari df_acc_clean.
    # Di Langkah 3.4.1 StandardScaler ikut men-scale issue_d_year (nunique > 2), sehingga
    # nilai di df_acc_clean sudah berupa z-score, bukan tahun sebenarnya. Snapshot menyimpan
    # nilai pra-scaling sehingga tahun yang diekspor tetap dalam satuan asli (2007-2018).
    _year_out = pd.DataFrame({
        'row_pos': np.arange(len(df_acc_final), dtype='int64'),
        'issue_year': snapshot_before_scale['issue_d_year'].to_numpy(),
    })
    if 'issue_d_month' in snapshot_before_scale:
        _year_out['issue_month'] = snapshot_before_scale['issue_d_month'].to_numpy()

    try:
        _year_out.to_parquet('../Datasets/Cleaning/Phase 2/issue_year_accepted.parquet',
                             index=False)
        _saved_as = 'issue_year_accepted.parquet'
    except Exception:  # pyarrow/fastparquet tidak tersedia -> jatuh ke CSV
        _year_out.to_csv('../Datasets/Cleaning/Phase 2/issue_year_accepted.csv', index=False)
        _saved_as = 'issue_year_accepted.csv'

    print(f'Saved: {_saved_as} -> {_year_out.shape} '
          f'(rentang tahun {int(_year_out["issue_year"].min())}-'
          f'{int(_year_out["issue_year"].max())})')
else:
    print('PERINGATAN: snapshot issue_d_year tidak ada; filter tahun tidak dapat diekspor.')

df_rej_clean.to_csv('../Datasets/Cleaning/Phase 2/clean_rejected_loans.csv', index=False)
print(f'Saved: clean_rejected_loans.csv -> {df_rej_clean.shape}')

# Verifikasi: pastikan fitur numerik df_rej benar-benar sudah ter-scale (mean~0, std~1)
print('\nVerifikasi normalisasi df_rej (fitur numerik untuk clustering):')
_check = df_rej_clean[numeric_for_scale_rej].agg(['mean', 'std']).round(4)
print(_check)
_ok = (_check.loc['mean'].abs() < 0.01).all() and ((_check.loc['std'] - 1).abs() < 0.05).all()
print('Status:', 'OK - df_rej sudah ternormalisasi (siap untuk clustering)' if _ok
      else 'PERIKSA - scaling df_rej belum sesuai')

print()
print('Semua file berhasil disimpan.')
```

### Ekspor Nilai Mentah Kolom Count untuk Verifikasi Anomali (Fase 4)

Karena StandardScaler menstandarisasi seluruh kolom numerik (termasuk acc_open_past_24mths
dan num_tl_op_past_12m), nilai yang tersimpan di clean_accepted_loans.csv berupa z-score,
bukan jumlah akun yang sebenarnya. Untuk Fase 4 saya perlu nilai mentah kedua kolom count
ini agar bisa memverifikasi apakah record ber-|z| sangat besar itu benar-benar error input
(nilai mustahil) atau sekadar count tinggi yang nyata. Saya sudah menyimpannya di
snapshot_before_scale; di sini saya mengekspornya ke file terpisah, disejajarkan dengan
posisi baris clean_accepted_loans.csv (index 0..N-1) agar pemetaan di Fase 4 tepat.

```python
# --- Ekspor nilai MENTAH count (pre-scaling) untuk verifikasi anomali Fase 4 ---
# df_acc_final disimpan ke CSV dengan index=False, jadi posisi baris pada CSV = 0..N-1
# mengikuti urutan df_acc_final. snapshot_before_scale diambil dari df_acc_clean dengan
# urutan baris yang SAMA (subset kolom tak mengubah urutan), sehingga bisa disejajarkan
# lewat posisi baris. Saya reset index agar 'row_pos' identik dengan cara Fase 4 membaca CSV.
count_raw_cols = [c for c in ['acc_open_past_24mths', 'num_tl_op_past_12m']
                  if c in snapshot_before_scale]

if count_raw_cols:
    raw_counts = pd.DataFrame(
        {c: snapshot_before_scale[c].values for c in count_raw_cols}
    )
    # Selaraskan panjang dengan df_acc_final (jaga-jaga bila ada perbedaan filter).
    if len(raw_counts) == len(df_acc_final):
        raw_counts = raw_counts.reset_index(drop=True)
        raw_counts.insert(0, 'row_pos', range(len(raw_counts)))
        try:
            out_raw = '../Datasets/Cleaning/Phase 2/raw_counts_accepted.parquet'
            raw_counts.to_parquet(out_raw, index=False)
            fmt = 'parquet'
        except Exception:
            out_raw = '../Datasets/Cleaning/Phase 2/raw_counts_accepted.csv'
            raw_counts.to_csv(out_raw, index=False)
            fmt = 'csv (fallback; pertimbangkan pip install pyarrow)'
        print(f'Nilai mentah count disimpan ({fmt}) -> {out_raw}')
        print(f'  {len(raw_counts):,} baris, kolom: {count_raw_cols}')
        print('\nStatistik nilai MENTAH (jumlah akun sebenarnya):')
        print(raw_counts[count_raw_cols].describe().loc[['min','max','mean','50%']].round(2).to_string())
    else:
        print(f'[peringatan] panjang snapshot ({len(raw_counts):,}) != df_acc_final '
              f'({len(df_acc_final):,}); ekspor dilewati agar tidak salah sejajar.')
else:
    print('[peringatan] kolom count tidak ada di snapshot_before_scale; ekspor dilewati.')
```
