```python
import pandas as pd

# 1. Load data asli Anda
df = pd.read_csv('clean_accepted_loan_without_scaling.csv')

# 2. Pilih fitur yang ingin disertakan
# Fitur yang sudah berbentuk kategori
cat_features = ['grade', 'sub_grade', 'term', 'int_rate_bin', 'dti_bin', 'loan_status', 'home_ownership']

# Fitur angka yang perlu di-binning (dikelompokkan) agar bisa masuk Apriori
num_features = ['last_fico_range_low', 'fico_range_low', 'loan_amnt', 'installment', 'annual_inc']

# 3. Proses Binning (Mengubah angka jadi kategori: Low, Medium, High)
df_clean = df[cat_features].copy()
for col in num_features:
    if col in df.columns:
        # Mengelompokkan angka menjadi 3 bagian sama besar
        df_clean[col+'_bin'] = pd.qcut(df[col], q=3, labels=['Low', 'Medium', 'High'], duplicates='drop')

# 4. One-Hot Encoding (Ubah kategori menjadi 0 dan 1)
# Inilah tahap akhir agar data "Siap Pakai" di Apriori
df_ready = pd.get_dummies(df_clean)

# Pastikan formatnya integer 1/0 agar ringan saat diproses
df_ready = df_ready.astype(int)

# 5. Simpan hasil cleaning
df_ready.to_csv('clean_for_apriori.csv', index=False)

print("Data sudah siap! Sekarang ada", df_ready.shape[1], "kolom biner yang bisa diproses Apriori.")
```
